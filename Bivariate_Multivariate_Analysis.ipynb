### Feature engineering

```python
zip_file_path = "Data/LFS_JV_CPI.csv.zip"
csv_file_name = "LFS_JV_CPI.csv"  

# Open and read the zipped CSV file
with zipfile.ZipFile(zip_file_path, "r") as z:
    with z.open(csv_file_name) as f:
        df = pd.read_csv(f)

df.head()
```

```python
df.info()
```

```python
# Calculating Labour force participation rate 
labour_force = df[df['lf_status'].isin([1, 2, 3])].groupby('REF_DATE')['5y_age'].sum().reset_index()
labour_force.rename(columns={'5y_age': 'labour_force'})

# Get total working-age population 
working_age_pop = df.groupby('REF_DATE')['5y_age'].sum()

# Calculate Labour Force Participation Rate (LFPR)
lf_participation = (labour_force / working_age_pop) * 100

LFPR_monthly = lf_participation.reset_index().rename(columns={'5y_age': 'LFPR_rate'})

```

```python
df1 = df.merge(LFPR_monthly, on="REF_DATE", how="left")

df1.head()
```

```python
df.info()
```

```python
# Calculating Labour force participation rate 
labour_force = df1[df1['lf_status'].isin([1, 2, 3])].groupby('REF_DATE')['5y_age'].sum().reset_index()
labour_force.rename(columns={'5y_age': 'labour_force'}, inplace=True)

# Filter unemployed individuals 
unemployed_df = df1[df1["lf_status"] == 3]

# Count unemployed per month
unemployed_per_month = unemployed_df.groupby("REF_DATE")['5y_age'].sum().reset_index()
unemployed_per_month.rename(columns={"5y_age": "Unemployed"}, inplace=True)

unemployment_rate_df = unemployed_per_month.merge(labour_force, on="REF_DATE")

# Compute unemployment rate
unemployment_rate_df["Unemployment_Rate"] = (unemployment_rate_df["Unemployed"] / unemployment_rate_df["labour_force"]) * 100

# Display results
print(unemployment_rate_df.head())
```

```python
# Merge unemployment rate into merged_df
df_final = df1.merge(unemployment_rate_df[["REF_DATE", "Unemployment_Rate"]], on="REF_DATE", how="left")
```
