## Preprocessing

Feature engineering labour force participation rate, unemployment rate, employment rate, and market tightness and merging with 'Total, All industries' Job vacancy data

```python
weight_col = 'final_wt'
# Define employment categories
employed = LFS[LFS['lf_status'].isin([1, 2])]
unemployed = LFS[LFS['lf_status'] == 3]
labour_force = LFS[LFS['lf_status'].isin([1, 2, 3])]


# Group by REF_DATE (monthly) and sum weights
monthly_employed = employed.groupby('REF_DATE')[weight_col].sum()
monthly_unemployed = unemployed.groupby('REF_DATE')[weight_col].sum()
monthly_labour_force = labour_force.groupby('REF_DATE')[weight_col].sum()
monthly_working_age_pop = LFS.groupby('REF_DATE')[weight_col].sum()

# Calculate Monthly Employment & Unemployment Rates
employment_rate = (monthly_employed / monthly_working_age_pop) * 100
unemployment_rate = (monthly_unemployed / monthly_labour_force) * 100
participation_rate = (monthly_labour_force / monthly_working_age_pop) * 100

# Combine into a DataFrame
monthly_rates = pd.DataFrame({
    'Employment Rate (%)': employment_rate,
    'Unemployment Rate (%)': unemployment_rate, 
    'Labour Force Participation Rate (%)': participation_rate,
    'Unemployed': monthly_unemployed
}).reset_index()

# Merging with JV & CPI data 
total_industries = JV_aggregated[JV_aggregated['combined_industries'] == 'Total, all industries']
monthly_rates = monthly_rates.merge(total_industries, on= 'REF_DATE', how='left')
monthly_rates = monthly_rates.merge(CPI_all_items, on= 'REF_DATE', how='left')


# Calculating Market Tightness 
monthly_rates["Market_Tightness"] = monthly_rates["Job Vacancies (#)"] / monthly_rates["Unemployed"]


monthly_rates
```

Feature engineering labour force participation rate, unemployment rate, employment rate, and market tightness by industry and merging with Job vacancy data

```python
# Grouping Employed (LF_STATUS: 1, 2) by month & industry
employed_by_month_industry = (
    LFS[LFS['lf_status'].isin([1, 2])]
    .groupby(['REF_DATE', 'NAICS_21'])[weight_col]
    .sum()
    .reset_index()
    .rename(columns={weight_col: 'total_employed'})
)

# Grouping Unemployed (LF_STATUS: 3) by month & industry
unemployed_by_month_industry = (
    LFS[LFS['lf_status'] == 3]
    .groupby(['REF_DATE', 'NAICS_21'])[weight_col]
    .sum()
    .reset_index()
    .rename(columns={weight_col: 'total_unemployed'})
)

# Merging employed & unemployed data
monthly_industry_rates = pd.merge(employed_by_month_industry, unemployed_by_month_industry, 
                                  on=['REF_DATE', 'NAICS_21'], how='outer')
monthly_industry_rates

# Labour Force per Industry per Month (Employed + Unemployed)
monthly_industry_rates['labour_force'] = (
    monthly_industry_rates['total_employed'] + monthly_industry_rates['total_unemployed']
)

# Calculating Monthly Employment & Unemployment Rates
monthly_industry_rates['employment_rate'] = (
    (monthly_industry_rates['total_employed'] / monthly_industry_rates['labour_force']) * 100
).round(2)

monthly_industry_rates['unemployment_rate'] = (
    (monthly_industry_rates['total_unemployed'] / monthly_industry_rates['labour_force']) * 100
).round(2)


monthly_industry_rates
```

```python
## Mapping JV industries to numerical categories

JV_aggregated.drop(JV_aggregated[JV_aggregated['combined_industries'] == 'Total, all industries'].index, inplace=True)

industry_encoding = {
    'Accommodation and food services [72]': 1, 
    'Business, building and other support services [55-56]': 2, 
    'Agriculture, forestry, fishing and hunting [11]': 3, 
    'Construction [23]': 4,
    'Educational services [61]': 5, 
    'Finance and insurance [52]': 6,
    'Health care and social assistance [62]': 7,
    'Information, culture and recreation [51-71]': 8, 
    'Manufacturing [31-33]': 9, 
    'Mining, quarrying, and oil and gas extraction [21]': 10, 
    'Other services (except public administration) [81]': 11,
    'Professional, scientific and technical services [54]': 12,
    'Public administration [91]': 13,
    'Real estate and rental and leasing [53]': 14,
    'Retail trade [44-45]': 15,
    'Transportation and warehousing [48-49]': 16,
    'Utilities [22]': 17,
    'Wholesale trade [41]': 18
}

JV_aggregated['JV_industry'] = JV_aggregated['combined_industries'].map(industry_encoding)

if JV_aggregated['JV_industry'].isnull().any():
    print("Warning: Some industries could not be mapped. Check for missing values.")

# Mapping categories in LFS to JV

LFS_to_JV = {
    1 : 3,
    2 : 3,  
    3 : 3,  
    4 : 10,
    5 : 17,
    6 : 4,
    7 : 9,  
    8 : 9,  
    9 : 18,
    10: 15,
    11: 16,
    12: 6,
    13: 14,
    14: 12,
    15: 2,  
    16: 5,
    17: 7,
    18: 8,  
    19: 1,
    20: 11,
    21: 13,
}

monthly_industry_rates['NAICS_21_JV'] = monthly_industry_rates['NAICS_21'].map(LFS_to_JV)

if monthly_industry_rates['NAICS_21_JV'].isnull().any():
    print("Warning: Some industries could not be mapped. Check for missing values.")

# Merging JV & CPI with industry monthly rates 
monthly_industry_rates = monthly_industry_rates.merge(JV_aggregated, left_on=["REF_DATE", "NAICS_21_JV"], right_on=["REF_DATE", "JV_industry"], how="left")
monthly_industry_rates = monthly_industry_rates.merge(CPI_all_items, on= 'REF_DATE', how='left') 
monthly_industry_rates
```

Agriculture, forestry, fishing and hunting [11] and Manufacturing [31-33] are each one industry in JV. However, in LFS the Manufacturing [31-33] is divided into 2 sub-industries and Agriculture, forestry, fishing and hunting [11] is divided into 3 sub-industries. The job vacancy data needs to be proportional to each industry. 

```python
adjusted_rates = monthly_industry_rates[monthly_industry_rates['NAICS_21'].isin([1,2,3,7,8])]

# Compute total labour force for each JV industry
total_labour_force_per_JV = (
    adjusted_rates.groupby(['REF_DATE', 'NAICS_21_JV'])['labour_force']
    .sum()
    .reset_index()
    .rename(columns={'labour_force': 'total_labour_force_JV'})
)

# Merge total labour force back into adjusted_rates
adjusted_rates = adjusted_rates.merge(total_labour_force_per_JV, on=['REF_DATE', 'NAICS_21_JV'], how='left')

# Compute each subindustry’s labour force proportion within the JV industry
adjusted_rates['labour_force_proportion'] = adjusted_rates['labour_force'] / adjusted_rates['total_labour_force_JV']

# Adjust Job Vacancies and Payroll Employees based on labour force proportion
adjusted_rates['Adjusted Job Vacancies'] = (
    adjusted_rates['Job Vacancies (#)'] * adjusted_rates['labour_force_proportion']
).round(0)  

adjusted_rates['Adjusted Payroll Employees'] = (
    adjusted_rates['Payroll Employees (#)'] * adjusted_rates['labour_force_proportion']
).round(0)  


adjusted_rates.drop(columns=['Job Vacancies (#)', 'Payroll Employees (#)'], inplace=True) 
adjusted_rates.rename(columns={'Adjusted Job Vacancies': 'Job Vacancies (#)', 'Adjusted Payroll Employees': 'Payroll Employees (#)'}, inplace=True)

# Industries that need to be updated
industries_to_update = [1, 2, 3, 7, 8]  # Agriculture & Manufacturing NAICS_21

# Filter out these industries from monthly industry rates dataset
monthly_industry_rates = monthly_industry_rates[~monthly_industry_rates['NAICS_21'].isin(industries_to_update)]

# Append the adjusted data back into the monthly industry rates dataset
updated_monthly_industry_rates = pd.concat([monthly_industry_rates, adjusted_rates], ignore_index=True)

# Ensure data is sorted correctly 
updated_monthly_industry_rates = updated_monthly_industry_rates.sort_values(by=['REF_DATE', 'NAICS_21']).reset_index(drop=True)

# Calculate Market Tightness 
updated_monthly_industry_rates["Market_Tightness"] = updated_monthly_industry_rates["Job Vacancies (#)"] / updated_monthly_industry_rates["total_unemployed"]

# Dropping redundant columns
updated_monthly_industry_rates.drop(columns=['total_employed', 'total_unemployed', 'total_labour_force_JV', 'labour_force_proportion', 
                                            'JV_industry', 'combined_industries', 'NAICS_21_JV', 'labour_force'], inplace=True)
updated_monthly_industry_rates
```
