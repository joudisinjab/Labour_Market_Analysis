### How do demographic factors such as age, gender, marital status, education level, and type of economic family influence labour force participation and unemployment rates?

```python
def calculate_lfpr(LFS, category):
    """
    Calculate the Labour Force Participation Rate (LFPR) by a given category.

    Parameters:
    LFS (pd.DataFrame): The DataFrame containing the labour force survey data.
    category (str): The column name for which the LFPR should be calculated.

    Returns:
    pd.DataFrame: A DataFrame with the LFPR by the given category.
    """
    # Count individuals in the labour force (lf_status 1, 2, 3) grouped by REF_DATE and the given category
    labour_force = LFS[LFS['lf_status'].isin([1, 2, 3])].groupby(['REF_DATE', category]).size().reset_index(name='labour_force')

    # Count total working-age population grouped by REF_DATE and the given category
    working_age_pop = LFS.groupby(['REF_DATE', category]).size().reset_index(name='working_age_pop')

    # Merge the two DataFrames
    lfp_df = labour_force.merge(working_age_pop, on=['REF_DATE', category], how='left')

    # Calculate Labour Force Participation Rate (LFPR)
    lfp_df['LFPR_rate'] = ((lfp_df['labour_force'] / lfp_df['working_age_pop']) * 100).round(2)

    # Drop unnecessary columns
    lfp_df.drop(columns=['labour_force', 'working_age_pop'], inplace=True)

    return lfp_df
```

```python
import pandas as pd

def calculate_unemployment(LFS, category):
    """
    Calculate the Unemployment Rate by a given category.

    Parameters:
    LFS (pd.DataFrame): The DataFrame containing the labour force survey data.
    category (str): The column name for which the Unemployment Rate should be calculated.

    Returns:
    pd.DataFrame: A DataFrame with the Unemployment Rate by the given category.
    """
    # Count individuals in the labour force (lf_status 1, 2, 3) grouped by REF_DATE and the given category
    labour_force = LFS[LFS['lf_status'].isin([1, 2, 3])].groupby(['REF_DATE', category]).size().reset_index(name='labour_force')

    # Filter unemployed individuals
    unemployed_df = LFS[LFS["lf_status"] == 3]

    # Count unemployed per month and category
    unemployed_per_month = unemployed_df.groupby(['REF_DATE', category]).size().reset_index(name="Unemployed")

    # Merge data to compute the unemployment rate
    unemployment_rate_df = unemployed_per_month.merge(labour_force, on=['REF_DATE', category], how="left")

    # Compute unemployment rate
    unemployment_rate_df["unemployment_rate"] = ((unemployment_rate_df["Unemployed"] / unemployment_rate_df["labour_force"]) * 100).round(2)

    # Drop unnecessary columns
    unemployment_rate_df.drop(columns=['Unemployed', 'labour_force'], inplace=True)

    return unemployment_rate_df
```

```python
categorical_vars = ['province', 'cmetrop_area', '5y_age', 'marital_stat', 'sex', 'education', 'immigration_stat', 'econ_fam_type'] 
lfpr_results = []

# Calculate LFPR by education category
for var in categorical_vars:
    lfpr_agg = calculate_lfpr(LFS, var)
    lfpr_results.append(lfpr_agg)
# Pivot the data for plotting with seaborn
    lfpr_agg_pivot = lfpr_agg.pivot(index='REF_DATE', columns=var, values='LFPR_rate').reset_index()

# Melt the pivoted DataFrame for seaborn plotting
    lfpr_agg_melted = pd.melt(lfpr_agg_pivot, id_vars=['REF_DATE'], var_name= var, value_name='LFPR_rate')

# Plot using seaborn
    plt.figure(figsize=(12, 6))
    sns.lineplot(data=lfpr_agg_melted, x='REF_DATE', y='LFPR_rate', hue=var, marker='o', palette='viridis')
    plt.title(f"Trend of Labour Force Participation Rate by {var} Over Time")
    plt.xlabel("Month")
    plt.ylabel("LFPR (%)")
    plt.xticks(rotation=45)
    plt.legend(title=var, bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.show()
```

```python
from statsmodels.formula.api import ols
# Perform ANOVA test
for var, lfpr_agg in zip(categorical_vars, lfpr_results):
    model = ols(f'LFPR_rate ~ C(var)', data=lfpr_agg).fit()
    anova_table = sm.stats.anova_lm(model, typ=2)
    print(anova_table)

# Interpret the results
    if anova_table['PR(>F)'][0] < 0.05:
        print(f"\nThere is a statistically significant difference in LFPR based on {var}.")
    else:
        print(f"\nThere is no statistically significant difference in LFPR based on {var}.")

# # Calculate Spearman correlation
# spearman_corr, p_value = stats.spearmanr(educ_agg['education'], educ_agg['LFPR_rate'])
# print(f"Spearman correlation: {spearman_corr}, p-value: {p_value}")

# # Plot heatmap of the correlation matrix
# plt.figure(figsize=(10, 8))
# sns.heatmap(educ_agg[['LFPR_rate', 'education']].corr(method='spearman'), annot=True, cmap='coolwarm', center=0)
# plt.title("Spearman Correlation Matrix of LFPR by Education Levels")
# plt.show()
```

Labour Force Participation Rate and Unemployment Rate are not normally distributed. Non-parametric tests will be employed to identify if there is significance between groups

```python
from scipy.stats import kruskal

categorical_vars = ['province', 'cmetrop_area', '5y_age', 'marital_stat', 'sex', 'education', 'immigration_stat', 'econ_fam_type'] 

# Function to perform Kruskal-Wallis test for each categorical variable
def kruskal_test(df, group_var, independent):
    groups = [df[df[group_var] == cat][independent] for cat in df[group_var].unique()]
    return kruskal(*groups)

# Run test on original categorical variables
kruskal_results_LFPR = {var: kruskal_test(LFS_inLF, var, 'LFPR_rate') for var in categorical_vars}
kruskal_results_UE = {var: kruskal_test(LFS_inLF, var, 'overall_unemp_Rate') for var in categorical_vars}

# Extract p-values
kruskal_pvalues_LFPR = {var: result.pvalue for var, result in kruskal_results_LFPR.items()}
kruskal_pvalues_UE = {var: result.pvalue for var, result in kruskal_results_UE.items()}

# Convert results into a DataFrame for better visualization
kruskal_results_summary = pd.DataFrame({
    'Variable': categorical_vars,
    'LFPR_p_value': [kruskal_pvalues_LFPR[var] for var in categorical_vars],
    'UE_p_value': [kruskal_pvalues_UE[var] for var in categorical_vars]
})

print(kruskal_results_summary)
```

```python
from itertools import combinations
from cliffs_delta import cliffs_delta
import pandas as pd

# List of categorical variables to analyze


# Metrics to analyze
metrics = ['LFPR_rate', 'overall_unemp_Rate']

# Store results
effect_size_results = []

# Loop through each categorical variable
for var in categorical_vars:
    unique_categories = LFS_inLF[var].dropna().unique()  # Get unique categories

    # Generate all unique category pairs for comparison
    for cat1, cat2 in combinations(unique_categories, 2):
        # Loop through each metric
        for metric in metrics:
            group1 = LFS_inLF[LFS_inLF[var] == cat1][metric]
            group2 = LFS_inLF[LFS_inLF[var] == cat2][metric]

            # Compute Cliff’s Delta
            d, interpretation = cliffs_delta(group1, group2)

            # Store the result
            effect_size_results.append({
                'Variable': var,
                'Category 1': cat1,
                'Category 2': cat2,
                'Metric': metric,
                'Effect Size': d,
                'Interpretation': interpretation
            })

# Convert to DataFrame
effect_size_df = pd.DataFrame(effect_size_results)

# Display results
print(effect_size_df['Interpretation'].unique())
```

While the Kruskal-Wallis test indicates significant differences between variables, and Dunn's post-hoc tests reveal some significant differences between groups, Cliff’s Delta—a non-parametric effect size measure—suggests that these differences are negligible. This implies that while statistical significance exists, the differences have no meaningful real-world impact.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

LFS_inLF = LFS.loc[(LFS['lf_status'] == 1) | (LFS['lf_status'] == 2) | (LFS['lf_status'] == 3)]

# Define features and target variables
X = LFS_inLF[categorical_vars]  # Independent variables
y_LFPR = LFS_inLF['LFPR_rate']  # Target for LFPR
y_UE = LFS_inLF['overall_unemp_Rate']  # Target for UE

# Train-test split
X_train, X_test, y_train_LFPR, y_test_LFPR = train_test_split(X, y_LFPR, test_size=0.2, random_state=42)
X_train, X_test, y_train_UE, y_test_UE = train_test_split(X, y_UE, test_size=0.2, random_state=42)

# Initialize and train Random Forest models
rf_LFPR = RandomForestRegressor(n_estimators=100, random_state=42)
rf_LFPR.fit(X_train, y_train_LFPR)

rf_UE = RandomForestRegressor(n_estimators=100, random_state=42)
rf_UE.fit(X_train, y_train_UE)

# Predictions
y_pred_LFPR = rf_LFPR.predict(X_test)
y_pred_UE = rf_UE.predict(X_test)

# Evaluate performance
def evaluate_model(y_true, y_pred, model_name):
    r2 = r2_score(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    print(f"{model_name} - R²: {r2:.3f}, RMSE: {rmse:.3f}, MAE: {mae:.3f}")

evaluate_model(y_test_LFPR, y_pred_LFPR, "LFPR Model")
evaluate_model(y_test_UE, y_pred_UE, "UE Model")

# Feature Importance
def plot_feature_importance(model, X, title):
    importance = pd.Series(model.feature_importances_, index=X.columns).sort_values(ascending=False)
    plt.figure(figsize=(10,6))
    sns.barplot(x=importance, y=importance.index, palette="viridis")
    plt.title(title)
    plt.xlabel('Feature Importance')
    plt.ylabel('Features')
    plt.show()

plot_feature_importance(rf_LFPR, X, "Feature Importance for LFPR")
plot_feature_importance(rf_UE, X, "Feature Importance for Unemployment Rate")
```

```python
from scipy.stats import mannwhitneyu, kstest, norm
# Separate data into two groups based on the sex
F_jobless = LFS_worked_past[LFS_worked_past['sex'] == 2]['duration_jobless']
M_jobless = LFS_worked_past[LFS_worked_past['sex'] == 1]['duration_jobless']

# Checking for normality
# Step 2: Check for normality using the Shapiro-Wilk test
ks_A = kstest(F_jobless, 'norm', args=(F_jobless.mean(), F_jobless.std()))
ks_B = kstest(M_jobless, 'norm', args=(M_jobless.mean(), M_jobless.std()))

print('The results of the Kolmogorov-Smirnov Normality test:')
print(f"Kolmogorov-Smirnov test for group A: D={ks_A.statistic}, p-value={ks_A.pvalue}")
print(f"Kolmogorov-Smirnov test for group B: D={ks_B.statistic}, p-value={ks_B.pvalue}")


if ks_A.pvalue > 0.05 and ks_B.pvalue > 0.05:
    print('The groups are normally distributed\n')
else: 
    print('The groups are not normally distributed\n')

    # Step 3: Visualize normality with Q-Q plots
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
stats.probplot(F_jobless, dist="norm", plot=plt)
plt.title('Q-Q Plot for Females')

plt.subplot(1, 2, 2)
stats.probplot(M_jobless, dist="norm", plot=plt)
plt.title('Q-Q Plot for Males')

plt.tight_layout()
plt.show()

# Perform the Wilcoxon rank-sum test (Mann-Whitney U test)
u_stat, p_value = mannwhitneyu(F_jobless, M_jobless, alternative='two-sided')

print('The results of the Wilcoxon rank-sum test:')
print(f"U-statistic: {u_stat}")
print(f"P-value: {p_value}\n")

# Determine if the result is statistically significant
alpha_1 = 0.05  # First significance level
alpha_2 = 0.01  # Second significance level

if p_value < alpha_2:
    print("The difference in means of duration of joblessness between Males and Females is highly statistically significant at the 0.01 level.")
elif p_value < alpha_1:
    print("The difference in means of duration of joblessness between Males and Females is statistically significant at the 0.05 level.")
else:
    print("The difference in means of duration of joblessness between Males and Females is not statistically significant at the 0.05 level.")
```

```python
groups = []
group_names = LFS_worked_past['education'].unique()

for group in LFS_worked_past['education'].unique():
    groups.append(LFS_worked_past[LFS_worked_past['education'] == group]['duration_jobless'])

# Step 3: Check for normality using the Kolmogorov-Smirnov test
for group_name, group in zip(group_names, groups):
    ks_stat, ks_pvalue = kstest(group, 'norm', args=(group.mean(), group.std()))
    print(f"Kolmogorov-Smirnov test for group {group_name}: D={ks_stat}, p-value={ks_pvalue}")
    if ks_pvalue > 0.05:
        print(f"Group {group_name} is normally distributed\n")
    else:
        print(f"Group {group_name} is not normally distributed\n")
        
# Step 3: Perform the Kruskal-Wallis test
stat, p_value = kruskal(*groups)

print('The results of the Kruskal Wallis Test:')
print(f"Kruskal-Wallis H-statistic: {stat}")
print(f"P-value: {p_value}")

# Determine if the result is statistically significant
alpha_1 = 0.05  # First significance level
alpha_2 = 0.01  # Second significance level

if p_value < alpha_2:
    print("The difference between groups is statistically significant at the 0.01 level.")
elif p_value < alpha_1:
    print("The difference between groups is statistically significant at the 0.05 level.")
else:
    print("The difference between groups is not statistically significant at the 0.05 level.")
```
